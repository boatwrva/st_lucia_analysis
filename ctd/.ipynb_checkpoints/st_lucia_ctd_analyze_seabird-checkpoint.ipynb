```python

def parse_cnv_file(filename):
    """Parse a Sea-Bird .cnv file into metadata and data."""
    metadata = {}
    data = []

    with open(filename, 'r') as file:
        for line in file:
            # Check for metadata (lines starting with '*')
            if line.startswith('*'):
                if '=' in line:
                    # Metadata key-value pair
                    key, value = line[1:].strip().split('=', 1)
                    metadata[key.strip()] = value.strip()
                else:
                    # Metadata comments without key-value
                    key = line[1:].strip()
                    metadata[key] = None
            elif line.startswith('#'):
                # Column headers start with '#'
                if 'name' in line:
                    _, col_index, col_name = line.strip().split(' ', 2)
                    metadata[f"Column_{col_index.strip()}"] = col_name.strip()
            else:
                # Data starts after metadata and column definitions
                data.append(line.strip())

    # Convert data into a DataFrame
    if data:
        data = [list(map(float, row.split())) for row in data if row]
        df = pd.DataFrame(data, columns=[metadata.get(f"Column_{i}", f"Column_{i}") for i in range(len(data[0]))])
    else:
        df = pd.DataFrame()

    return metadata, df
```

```python
## for kristina - she needs a csv file of each station's temp, fluoro, and do 
# at 5m, 100m, 1000m, and bottom 

# station_depths = [0 = 50, 1= 3850, 2 = 3890, 3= 3878, 4= 1248, 5= 4074, 6=4004 , 7=1653 , 8= 520, 9= 3703, 10 = 3815, 11=1615, 12=586]

station_depths = [50, 3850, 4074, 3703, 1615, 1653, 3890, 4004, 3815, 586, 520, 1248, 3878]
len(station_depths)

def extract_depth_profiles(survey_data, stations, station_depths, cruise, target_depths=[5, 100, 1000]):
    """
    Extract temperature, chlorophyll (fluorometer), and oxygen at target depths
    plus the nearest-to-bottom sample for each station.
    
    Parameters:
        survey_data: np.array of shape (num_stations, num_variables, deepest)
        stations: list of station indices (length = num casts, index 0 is wildfire station for dante)
        station_depths: list of bottom depths per cast (same length as castnumber)
        cruise: string cruise ID
        target_depths: list of target pressure levels (default [5, 100, 1000])
    
    Variable indices as saved in cnv files:
        0: Temp1, 2: Pressure, 6: Fluorometer (chl proxy), 9: Oxygen
    """

    # based on victoria's sle processing 
    VAR_TEMP     = 0
    VAR_PRESSURE = 2
    VAR_FLUORO   = 6  
    VAR_OXYGEN   = 9

    castnumber = np.arange(1, 14)  # casts 1-13
    rows = []

    for ii, cc in enumerate(castnumber):
        ss = stations[ii]  # station index into survey_data

        # Pull this station's pressure and variable vectors (strip trailing NaNs)
        pressure = survey_data[ss, VAR_PRESSURE, :]
        valid_mask = ~np.isnan(pressure)
        pressure = pressure[valid_mask]

        if len(pressure) == 0:
            print(f"  Cast {cc} (station index {ss}): no valid pressure data, skipping.")
            continue

        temp    = survey_data[ss, VAR_TEMP,   valid_mask]
        fluoro  = survey_data[ss, VAR_FLUORO,  valid_mask]
        oxygen  = survey_data[ss, VAR_OXYGEN,  valid_mask]

        row = {
            'cast':          cc,
            'station_index': ss,
            'bottom_depth_m': station_depths[ii],
        }
        
        # --- Quality control: mask out-of-water / surface noise ---
        # Salinity < 30 psu indicates the CTD was not yet submerged (or noisy surface hit)
        # NaN out ALL variables at those samples so they don't contaminate extractions
        sal1 = survey_data[ss, 12, valid_mask]
        bad_sal = sal1 < 30
        
        temp[bad_sal]   = np.nan
        fluoro[bad_sal] = np.nan
        oxygen[bad_sal] = np.nan
        pressure[bad_sal] = np.nan  # also mask pressure so these samples are invisible to argmin
        
        # Rebuild a clean mask after sal QC (re-strip the newly NaN'd pressure values)
        qc_mask = ~np.isnan(pressure)
        pressure = pressure[qc_mask]
        temp     = temp[qc_mask]
        fluoro   = fluoro[qc_mask]
        oxygen   = oxygen[qc_mask]

        
        # --- Target depths (5, 100, 1000 m) ---
        for td in target_depths:
            label = f'{td}m'
            
            if td == 5:
                # For the surface value, just take the shallowest good sample
                # (after sal QC, pressure[0] is the first clean in-water measurement)
                if len(pressure) == 0:
                    print(f"  Cast {cc}: no valid surface data after QC — filling NaN")
                    row[f'pressure_{label}']    = np.nan
                    row[f'temperature_{label}'] = np.nan
                    row[f'chlorophyll_{label}'] = np.nan
                    row[f'oxygen_{label}']      = np.nan
                    continue
                idx = 0  # shallowest clean sample
                actual_p = pressure[idx]
            else:
                    
                # Only search the DOWNCAST: find the first index where pressure exceeds
                # the target depth, then look in a small window around it
                exceeds_target = np.where(pressure >= td)[0]
                
                if len(exceeds_target) == 0:
                    # Station is shallower than this target depth
                    print(f"  Cast {cc}: station shallower than {td} m — filling NaN")
                    row[f'pressure_{label}']     = np.nan
                    row[f'temperature_{label}']  = np.nan
                    row[f'chlorophyll_{label}']  = np.nan
                    row[f'oxygen_{label}']       = np.nan
                    continue
                
                # First crossing of the target depth = start of downcast at that depth
                first_crossing = exceeds_target[0]
                
                # Look in a small window just around the first crossing to find
                # the nearest sample (in case sampling isn't exactly on the meter)
                window = 10  # samples on either side
                window_start = max(0, first_crossing - window)
                window_end   = min(len(pressure), first_crossing + window)
                
                idx = window_start + np.argmin(np.abs(pressure[window_start:window_end] - td))
                actual_p = pressure[idx]
            
            
            # Only fill if we actually got close (within 20 m for shallow, 50 m for deep)
            tolerance = 20 if td <= 100 else 50
            if np.abs(actual_p - td) <= tolerance:
                row[f'pressure_{label}']     = actual_p
                row[f'temperature_{label}']  = temp[idx]
                row[f'chlorophyll_{label}']  = fluoro[idx]
                row[f'oxygen_{label}']       = oxygen[idx]
            else:
                print(f"  Cast {cc}: no sample within {tolerance} m of {td} m "
                      f"(closest = {actual_p:.1f} m) — filling NaN")
                row[f'pressure_{label}']     = np.nan
                row[f'temperature_{label}']  = np.nan
                row[f'chlorophyll_{label}']  = np.nan
                row[f'oxygen_{label}']       = np.nan

        # --- Near-bottom sample ---
        # Use the altimeter (column 7) if available; otherwise use deepest valid sample
        altimeter = survey_data[ss, 7, :]
        
        altimeter = altimeter[valid_mask]   # first mask (NaN pressure)
        altimeter = altimeter[qc_mask]      # second mask (sal QC) — now same length as pressure

        
        # Strategy: find samples where altimeter > 0 (active return = near bottom)
        # and take the one with the highest pressure (deepest)
        alt_valid = altimeter > 0
        if alt_valid.any():
            # Among altimeter returns, pick deepest
            bottom_candidates = np.where(alt_valid)[0]
            idx_bottom = bottom_candidates[np.argmax(pressure[bottom_candidates])]
            method = 'altimeter'
        else:
            # Fall back to simply the deepest sample
            idx_bottom = np.argmax(pressure)
            method = 'deepest_sample'

        row['pressure_bottom']      = pressure[idx_bottom]
        row['temperature_bottom']   = temp[idx_bottom]
        row['chlorophyll_bottom']   = fluoro[idx_bottom]
        row['oxygen_bottom']        = oxygen[idx_bottom]
        row['bottom_method']        = method

        rows.append(row)
        print(f"  Cast {cc:02d} (station {ss}): "
              f"bottom @ {pressure[idx_bottom]:.1f} m via {method}")

    # Build DataFrame with a logical column order
    df_out = pd.DataFrame(rows)
    
    # Reorder columns cleanly
    id_cols    = ['cast', 'station_index', 'bottom_depth_m']
    data_cols  = []
    for label in [f'{d}m' for d in target_depths] + ['bottom']:
        for var in ['pressure', 'temperature', 'chlorophyll', 'oxygen']:
            data_cols.append(f'{var}_{label}')
    data_cols.append('bottom_method')
    
    df_out = df_out[id_cols + data_cols]
    return df_out


# run function 
print("Extracting depth profiles...")
df_summary = extract_depth_profiles(
    survey_data,
    stations,
    station_depths,
    cruise,
    target_depths=[5, 100, 1000]
)

print("\nResult:")
print(df_summary.to_string(index=False))

# Save to CSV
outfile = f'{cruise}_depth_profile_summary.csv'
df_summary.to_csv(outfile, index=False, float_format='%.4f')
print(f"\nSaved to {outfile}")
```

```python
def plot_profiles_with_extracted_depths(survey_data, stations, df_summary, cruise, save_loc=None):

    castnumber = np.arange(2, 14)  # skip cast 1 (wildfire station)
    target_labels = ['5m', '100m', '1000m', 'bottom']
    star_colors   = ['gold', 'orange', 'red', 'darkred']

    for ii, cc in enumerate(castnumber):
        ss = stations[ii + 1]  # +1 to skip wildfire station

        # Pull this cast's row from df_summary — this is what we are validating 
        cast_row = df_summary[df_summary['cast'] == cc]
        if cast_row.empty:
            print(f"Cast {cc} not found in summary, skipping.")
            continue
        cast_row = cast_row.iloc[0]

        # Full raw profiles for background lines only
        pressure = survey_data[ss, 2, :]
        temp     = survey_data[ss, 0, :]
        sal      = survey_data[ss, 12, :]
        fluoro   = survey_data[ss, 6, :]
        oxygen   = survey_data[ss, 9, :]

        ymax = np.nanmax(pressure) * 1.05

        fig, axes = plt.subplots(1, 4, figsize=(10, 5), sharey=True)
        fig.suptitle(f'{cruise} — Cast {cc:02d}  (station #{ss})',
                     fontsize=13, fontweight='bold', y=0.95)

        var_info = [
            (axes[0], temp,   'Temperature [°C]',       'Reds',    'temperature'),
            (axes[1], sal,    'Salinity [psu]',          'Purples', 'salinity'),
            (axes[2], fluoro, 'Fluorescence [spt]',      'Greens',  'chlorophyll'),
            (axes[3], oxygen, 'Dissolved Oxygen [mg/L]', 'Blues',   'oxygen'),
        ]

        for ax, profile, xlabel, cmap_name, var_key in var_info:
            # Full profile line (raw survey_data, just for visual background)
            color = plt.get_cmap(cmap_name)(0.7)
            ax.plot(profile, pressure, color=color, lw=1.2, alpha=0.85)

            # Stars come DIRECTLY from df_summary columns
            for label, sc in zip(target_labels, star_colors):
                p_col = f'pressure_{label}'
                v_col = f'{var_key}_{label}'

                # Make sure the column exists (e.g. salinity_bottom won't exist)
                if p_col not in cast_row.index or v_col not in cast_row.index:
                    continue

                p_val = cast_row[p_col]
                v_val = cast_row[v_col]

                if pd.isna(p_val) or pd.isna(v_val):
                    continue

                ax.scatter(v_val, p_val,
                           marker='*', s=180, color=sc,
                           edgecolors='k', linewidths=0.5,
                           zorder=5, label=f'{label} ({p_val:.0f} m)')

            ax.set_ylim([ymax, 0])
            ax.set_xlabel(xlabel, fontsize=10)
            ax.grid(True, alpha=0.3)
            ax.tick_params(axis='both', labelsize=9)

        axes[0].set_ylabel('Pressure [dbar]', fontsize=10)

        sal_data = sal[~np.isnan(sal)]
        if len(sal_data) > 0:
            axes[1].set_xlim([np.nanmin(sal_data) - 0.1, np.nanmax(sal_data) + 0.1])

        handles, lbls = axes[0].get_legend_handles_labels()
        if handles:
            fig.legend(handles, lbls,
                       loc='lower center', ncol=4, fontsize=9,
                       title='Sample depth', title_fontsize=9,
                       bbox_to_anchor=(0.5, -0.06), frameon=True)

        plt.tight_layout()

        if save_loc:
            fname = f'{save_loc}{cruise}_station{ss:02d}_profile_extracted_depths.png'
            fig.savefig(fname, dpi=200, bbox_inches='tight')
            print(f"Saved: {fname}")

        plt.show()
        plt.close()



# --- Run it ---
save_loc = '/home/vboatwright/OneDrive/Documents/SIO/projects/santalucia/figures/forKristina/'

plot_profiles_with_extracted_depths(
    survey_data,
    stations,
    df_summary,
    cruise,
    save_loc=save_loc   # set to None to just display without saving
)
```

```python
def plot_profiles_with_extracted_depths(survey_data, stations, df_summary, cruise, save_loc=None):
    """
    For each cast, plot the full vertical profile of temp, salinity, fluorescence, 
    and oxygen, with the extracted depths (5m, 100m, 1000m, bottom) marked as stars.
    """
    
    castnumber = np.arange(2, 14)  # skip cast 1 (wildfire station)
    target_labels = ['5m', '100m', '1000m', 'bottom']
    star_colors = ['gold', 'orange', 'red', 'darkred']
    
    for ii, cc in enumerate(castnumber):
        ss = stations[ii + 1]  # +1 to skip wildfire station index
        
        # Pull this cast's row from the summary DataFrame
        cast_row = df_summary[df_summary['cast'] == cc]
        if cast_row.empty:
            print(f"Cast {cc} not found in summary, skipping.")
            continue
        cast_row = cast_row.iloc[0]
        
        # Full profiles
        pressure = survey_data[ss, 2, :]
        temp     = survey_data[ss, 0, :]
        sal      = survey_data[ss, 12, :]
        fluoro   = survey_data[ss, 6, :]
        oxygen   = survey_data[ss, 9, :]
        
        ymax = np.nanmax(pressure) * 1.05  # a little breathing room below bottom
        
        fig, axes = plt.subplots(1, 4, figsize=(14, 7), sharey=True)
        fig.suptitle(f'{cruise} — Cast {cc:02d}  (station index {ss})', 
                     fontsize=13, fontweight='bold', y=1.01)
        
        var_info = [
            (axes[0], temp,   'Temperature [°C]',      'Reds',   'temperature'),
            (axes[1], sal,    'Salinity [psu]',         'Purples','salinity'),
            (axes[2], fluoro, 'Fluorescence [spt]',     'Greens', 'chlorophyll'),
            (axes[3], oxygen, 'Dissolved Oxygen [mg/L]','Blues',  'oxygen'),
        ]
        
        for ax, profile, xlabel, cmap_name, var_key in var_info:
            # Full profile line
            color = plt.get_cmap(cmap_name)(0.7)
            ax.plot(profile, pressure, color=color, lw=1.2, alpha=0.85)
            
            # Overlay extracted depth stars
            for label, sc in zip(target_labels, star_colors):
                p_val = cast_row.get(f'pressure_{label}', np.nan)
                v_val = cast_row.get(f'{var_key}_{label}', np.nan)
                
                if np.isnan(p_val) or np.isnan(v_val):
                    continue
                
                ax.scatter(v_val, p_val, 
                           marker='*', s=180, color=sc, 
                           edgecolors='k', linewidths=0.5,
                           zorder=5, label=f'{label} ({p_val:.0f} m)')
            
            ax.set_ylim([ymax, 0])
            ax.set_xlabel(xlabel, fontsize=10)
            ax.grid(True, alpha=0.3)
            ax.tick_params(axis='both', labelsize=9)
        
        # Y-axis label only on leftmost panel
        axes[0].set_ylabel('Pressure [dbar]', fontsize=10)
        
        # Salinity x-limits
        sal_data = sal[~np.isnan(sal)]
        if len(sal_data) > 0:
            axes[1].set_xlim([np.nanmin(sal_data) - 0.1, np.nanmax(sal_data) + 0.1])
        
        # Shared legend from first axis (stars are same across all panels)
        handles, lbls = axes[0].get_legend_handles_labels()
        if handles:
            fig.legend(handles, lbls, 
                       loc='lower center', ncol=4, fontsize=9,
                       title='Extracted depths', title_fontsize=9,
                       bbox_to_anchor=(0.5, -0.06), frameon=True)
        
        plt.tight_layout()
        
        if save_loc:
            fname = f'{save_loc}{cruise}_Station{ss:02d}_profile_extracted_depths.png'
            fig.savefig(fname, dpi=200, bbox_inches='tight')
            print(f"Saved: {fname}")
        
        plt.show()
        plt.close()


# --- Run it ---
save_loc = '/home/vboatwright/OneDrive/Documents/SIO/projects/santalucia/figures/cruise_report_figures/'

plot_profiles_with_extracted_depths(
    survey_data,
    stations,
    df_summary,
    cruise,
    save_loc=save_loc   # set to None to just display without saving
)
```

```python
# next, process in a loop to plot all 

path = '/media/vboatwright/KBZ/SR2503_scienceparty_share/CTD/'
deploy = 'PO_Stations/station_2/'
cruise = 'SR2503'; station_number = 2; cast_number = 1; 
filename = f'{cruise}_POstation_{station_number}_cast_{cast_number:02d}_updown.cnv'

# loop through each cast at the station 

files = set(os.listdir(path+deploy))
cnv_files = [f for f in files if fnmatch.fnmatch(f, '*.cnv')]
num_casts = len(cnv_files)


num_variables = 17
variable_order = ['Temp','Cond','Pressure','Temp2','Cond2','Transmissometer','Fluorometer','Alimeter','PAR','Oxygen','UPV','SPAR','Salinity','Salinity2','Lon','Lat','Flag']
# longest (?) shape of p/t/c object: 212103
deepest = 160000
postation_data = np.zeros((2,num_casts,num_variables,deepest))*np.nan

# loop through PO station 1 casts 
station_number = 1

# for big concatenations 
p_all1 = []
temp1_all1 = []
sal1_all1 = []
oxy_all1 = []
chl_all1 = []

p_all2 = []
temp1_all2 = []
sal1_all2 = []
oxy_all2 = []
chl_all2 = []

for ss in np.arange(0,2): 
    # loop through PO stations 
    
    cruise = 'SR2503'; 
    station_number = ss+1;  
    deploy = f'PO_Stations/station_{station_number}/'
    files = set(os.listdir(path+deploy))
    cnv_files = [f for f in files if fnmatch.fnmatch(f, '*.cnv')]
    num_casts = len(cnv_files)
        
    for ii in np.arange(0,num_casts):
        filename = f'{cruise}_POstation_{station_number}_cast_{ii+1:02d}_updown.cnv'
        metadata, data = parse_cnv_file(path+deploy+filename)
    
        zlen = len(data['Column_2'].values)
        
        temp1 = data['Column_0'].values
        postation_data[ss,ii,0,0:zlen] = temp1
        
        cond1 = data['Column_1'].values
        postation_data[ss,ii,1,0:zlen] = cond1
    
        p = data['Column_2'].values
        postation_data[ss,ii,2,0:zlen] = p
    
        temp2 = data['Column_3'].values
        postation_data[ss,ii,3,0:zlen] = temp2
    
        cond2 = data['Column_4'].values
        postation_data[ss,ii,4,0:zlen] = cond2
        
        trans = data['Column_5'].values
        postation_data[ss,ii,5,0:zlen] = trans
        
        fluoro = data['Column_6'].values
        postation_data[ss,ii,6,0:zlen] = fluoro
        
        altimeter = data['Column_7'].values
        postation_data[ss,ii,7,0:zlen] = altimeter
    
        PAR = data['Column_8'].values
        postation_data[ss,ii,8,0:zlen] = PAR
        
        oxygen = data['Column_9'].values
        postation_data[ss,ii,9,0:zlen] = oxygen
        
        upv = data['Column_10'].values
        postation_data[ss,ii,10,0:zlen] = upv
        
        SPAR = data['Column_11'].values
        postation_data[ss,ii,11,0:zlen] = SPAR
        
        sal1 = data['Column_12'].values
        postation_data[ss,ii,12,0:zlen] = sal1
        
        sal2 = data['Column_13'].values
        postation_data[ss,ii,13,0:zlen] = sal2
        
        lon = data['Column_14'].values
        postation_data[ss,ii,14,0:zlen] = lon
        
        lat = data['Column_15'].values
        postation_data[ss,ii,15,0:zlen] = lat
    
        flag = data['Column_16'].values
        postation_data[ss,ii,16,0:zlen] = flag
    
        if zlen > 150000: 
            z_coord = p
    
        # append them all 
        if ss == 0: 
                
            p_all1 = np.append(p_all1,p)
            temp1_all1 = np.append(temp1_all1,temp1)
            sal1_all1 = np.append(sal1_all1,sal1)
            oxy_all1 = np.append(oxy_all1,oxygen)
            chl_all1 = np.append(chl_all1,fluoro)

        if ss == 1: 
            
            p_all2 = np.append(p_all2,p)
            temp1_all2 = np.append(temp1_all2,temp1)
            sal1_all2 = np.append(sal1_all2,sal1)
            oxy_all2 = np.append(oxy_all2,oxygen)
            chl_all2 = np.append(chl_all2,fluoro)

```
